# Clinical Data Visualization Notebook

# This notebook contains:
 - Clinical KPI visualizations
 - Injury type comparisons
 - Patient outcome charts
 - Correlation plots
 - Advanced healthcare visual analytics



```python
# Imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
sns.set(style="whitegrid")

```

```python
# Load Cleaned Dataset
df = pd.read_csv("../data/clinical_data_cleaned.csv")
df.head()

```

## SECTION 1 — KPI Visualizations

```python
# Recovery Days Distribution
plt.figure(figsize=(10, 6))
sns.histplot(df["recovery_days"], kde=True, color="blue")
plt.title("Recovery Days Distribution")
plt.xlabel("Days")
plt.ylabel("Count")
plt.show()

```

```python
# Pain Improvement Distribution
plt.figure(figsize=(10, 6))
sns.histplot(df["pain_improvement"], kde=True, color="red")
plt.title("Pain Improvement Distribution")
plt.xlabel("Pain Reduction")
plt.ylabel("Count")
plt.show()

```

```python
# Mobility Gain Distribution
plt.figure(figsize=(10, 6))
sns.histplot(df["mobility_gain"], kde=True, color="green")
plt.title("Mobility Gain Distribution")
plt.xlabel("Mobility Gain")
plt.ylabel("Count")
plt.show()

```

## SECTION 2 — Injury Type Visualizations

```python
# Recovery Days by Injury Type
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x="injury_type", y="recovery_days", palette="Blues")
plt.title("Average Recovery Days by Injury Type")
plt.xlabel("Injury Type")
plt.ylabel("Recovery Days")
plt.show()

```

```python
# Success Rate by Injury Type
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x="injury_type", y="success", palette="Greens")
plt.title("Success Rate by Injury Type")
plt.xlabel("Injury Type")
plt.ylabel("Success Rate")
plt.show()

```

```python
# Pain Improvement by Injury Type
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x="injury_type", y="pain_improvement", palette="Reds")
plt.title("Pain Improvement by Injury Type")
plt.xlabel("Injury Type")
plt.ylabel("Pain Improvement")
plt.show()

```

```python
# Mobility Gain by Injury Type
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x="injury_type", y="mobility_gain", palette="Greens")
plt.title("Mobility Gain by Injury Type")
plt.xlabel("Injury Type")
plt.ylabel("Mobility Gain")
plt.show()

```

## SECTION 3 — Correlations & Relationships

```python
# Correlation Matrix
plt.figure(figsize=(10, 6))
sns.heatmap(df[["recovery_days", "pain_improvement", "mobility_gain", "sessions"]].corr(),
            annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()

```

```python
# Mobility Gain vs Pain Improvement
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x="pain_improvement", y="mobility_gain", hue="injury_type", palette="Set2")
plt.title("Mobility Gain vs Pain Improvement")
plt.xlabel("Pain Improvement")
plt.ylabel("Mobility Gain")
plt.show()

```

```python
# Sessions vs Recovery Days
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x="sessions", y="recovery_days", hue="injury_type", palette="Set1")
plt.title("Sessions vs Recovery Days")
plt.xlabel("Sessions")
plt.ylabel("Recovery Days")
plt.show()

```

## SECTION 4 — Patient Evolution Visualizations

```python
# Pain Evolution per Patient
plt.figure(figsize=(12, 6))
sns.lineplot(data=df, x="patient_id", y="pain_score_initial", label="Initial Pain", color="red")
sns.lineplot(data=df, x="patient_id", y="pain_score_final", label="Final Pain", color="green")
plt.title("Pain Evolution per Patient")
plt.xlabel("Patient ID")
plt.ylabel("Pain Score")
plt.legend()
plt.show()

```

```python
# Mobility Evolution per Patient
plt.figure(figsize=(12, 6))
sns.lineplot(data=df, x="patient_id", y="mobility_initial", label="Initial Mobility", color="orange")
sns.lineplot(data=df, x="patient_id", y="mobility_final", label="Final Mobility", color="blue")
plt.title("Mobility Evolution per Patient")
plt.xlabel("Patient ID")
plt.ylabel("Mobility Score")
plt.legend()
plt.show()
```
