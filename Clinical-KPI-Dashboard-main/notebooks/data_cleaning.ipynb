# Clinical Data Cleaning & KPI Generation Notebook

# This notebook performs:
- Data ingestion
- Cleaning and validation
- KPI calculations
- Clinical metrics generation
- Export of cleaned dataset


```python
# Imports
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
```

```python
# Load Dataset
# Load synthetic clinical dataset
df = pd.read_csv("../data/clinical_data.csv")

df.head()
```

```python
# Basic Inspection
df.info()
df.describe()
df.isnull().sum()
```

```python
# Remove Duplicates
df = df.drop_duplicates()
print("Duplicates removed.")
```

```python
# Convert Dates
df["start_date"] = pd.to_datetime(df["start_date"])
df["end_date"] = pd.to_datetime(df["end_date"])
```

```python
# Create KPIs
# Recovery time
df["recovery_days"] = (df["end_date"] - df["start_date"]).dt.days

# Pain improvement
df["pain_improvement"] = df["pain_score_initial"] - df["pain_score_final"]

# Mobility gain
df["mobility_gain"] = df["mobility_final"] - df["mobility_initial"]

# Success flag
df["success"] = (df["mobility_gain"] > 20) & (df["pain_improvement"] > 2)

df.head()
```

```python
# Validate KPI Ranges
print("Recovery days:", df["recovery_days"].describe())
print("Pain improvement:", df["pain_improvement"].describe())
print("Mobility gain:", df["mobility_gain"].describe())
print("Success rate:", df["success"].mean())
```

```python
# Injury Type Summary
df.groupby("injury_type")[["recovery_days", "pain_improvement", "mobility_gain", "success"]].mean()
```

```python
# Export Cleaned Dataset
df.to_csv("../data/clinical_data_cleaned.csv", index=False)
print("Cleaned dataset exported to /data/clinical_data_cleaned.csv")
```
